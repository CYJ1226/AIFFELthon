### 평가 시작

````python
import pandas as pd
from datasets import Dataset
from ragas import evaluate
from ragas.metrics import faithfulness, answer_relevancy, context_precision, context_recall, answer_similarity
from langchain_openai import ChatOpenAI
````

````python
# results 형변환
df = pd.DataFrame(evals)
ragas_dataset = Dataset.from_pandas(df)
````

````python
df
````

````python
# # ragas 최신버전은 Langchain LLM 객체 형태를 받는 구조로 설계됨
# # Invalid JSON 에러 발생하면 model_kwargs={"response_format": {"type": "json_object"} 매개변수 ChatOpenAI에 추가
judge_llm = ChatOpenAI(model='gpt-4.1-mini', temperature=0) # 4.1 mini가 가격, 성능 대비 가장 최적, 없는 모델이면 invoke시 오류 발생하여 테스트 가능

faithfulness.llm = judge_llm
answer_relevancy.llm = judge_llm
````

````python
metrics = [
    faithfulness,
    answer_relevancy,
]
````

````python
score = evaluate(ragas_dataset, metrics=metrics)
````

````python
score # chat: 5.1, judge: 4.1-mini
````

## Opik에 trace 생성

````python
# 프로젝트 이름
# proj_name = "Base (gpt-4.1)"
# proj_name = "MultiQuery (gpt-4.1)"
# proj_name = "Reranker (gpt-4.1)"
# proj_name = "Multiquery(gpt-5.1)"
# proj_name = "Reranker(gpt-5.1)"
proj_name = "audit_rag7"
````

````python
for i in range(len(df)):
    client.trace(
        project_name=proj_name,
        input={"question": df['question'][i]},
        output={"answer": df['answer'][i]},
        metadata={"contexts": df['contexts'][i]},
        feedback_scores=[{"name": "Faithfulness (Ragas)", "value": score['faithfulness'][i]},
                         {"name": "Answer_Relevancy (Ragas)", "value": score['answer_relevancy'][i]}]
        # tags=[]
    )
````
